### 2. Data

In PyTorch, data is usually prepared before it is passed into a neural network.

Two important tools for this are `Dataset` and `DataLoader`.

#### Dataset

A `Dataset` defines how to get data samples and their labels.

The main job of a dataset is to answer this question:

> If I give you an index, what data sample and label should you return?

For example, in image classification, one dataset item may contain one image and its class label.

#### DataLoader

A `DataLoader` takes data from a `Dataset` and feeds it to the model in a more convenient way.

It is commonly used to:

- group samples into batches;
- shuffle the data;
- load data during training.

In short, `Dataset` defines how to access one sample, and `DataLoader` defines how to deliver many samples to the model.

```python
# initialization
root_dir="datasets/hymenoptera_data/train"
label_dir="ants"
path=os.path.join(root_dir,label_dir)
```

```python
class MyData(Dataset):
    def __init__(self,root_dir,label_dir):
        self.root_dir=root_dir
        self.label_dir=label_dir
        self.path=os.path.join(self.root_dir,self.label_dir)
        self.img_path=os.listdir(self.path)

    def __getitem__(self,idx):
        img_name=self.img_path[idx]
        img_item_path=os.path.join(self.root_dir,self.label_dir,img_name)
        img=Image.open(img_item_path)
        label=self.label_dir
        return img,label
    
    def __len__(self):
        return len(self.img_path)
```

```python
# Example 
root_dir="datasets/hymenoptera_data/train"
ants_label_dir="ants"
bees_label_dir="bees"
ants_dataset=MyData(root_dir,ants_label_dir)
bees_dataset=MyData(root_dir,bees_label_dir)

train_dataset=ants_dataset + bees_dataset
train_dataset[0]
```
