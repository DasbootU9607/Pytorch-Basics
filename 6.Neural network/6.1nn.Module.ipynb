### `nn.Module` Basics

`nn.Module` is the base class for neural network models in PyTorch. We define layers in `__init__()` and describe the data flow in `forward()`.

#### 1. Forward Pass Idea

A common pattern is: input data goes through layers, activation functions add non-linearity, and the final output is returned.

```python
def forward(self, x):
    x = F.relu(self.conv1(x))
    return F.relu(self.conv2(x))
```

This example means: apply `conv1`, apply `ReLU`, then apply `conv2` and `ReLU` again.

#### 2. Build A Simple Module

This small module does not have trainable layers. It only adds `1` to the input so we can see how `forward()` works.

````python
import torch
from torch import nn
````

````python
class Example(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, input):
        # The forward method defines what happens when the module is called.
        output = input + 1
        return output
````

#### 3. Use The Module

Calling `eg(x)` automatically runs `eg.forward(x)`.

````python
eg = Example()
x = torch.tensor(1.0)

output = eg(x)
print(output)
````

#### Key Idea

Use `nn.Module` when you want to build a reusable model or layer. Put setup in `__init__()` and computation in `forward()`.
