### Convolution Layers

A convolution layer is commonly used for image data. It looks at small local regions of an image and learns filters for useful patterns such as edges, colors, or textures.

#### 1. Imports

`torch` creates tensors. `nn` provides neural network layers such as `Conv2d`.

```python
import torch
from torch import nn
```

#### 2. Input Shape

For `Conv2d`, image tensors use this shape: `[batch_size, channels, height, width]`.

Example: CIFAR-10 images have 3 color channels and size `32 x 32`.

```python
# Create a fake batch of 4 RGB images.
x = torch.randn(4, 3, 32, 32)
print(x.shape)
```

#### 3. Create A Convolution Layer

`in_channels=3` matches RGB input images. `out_channels=6` means the layer learns 6 filters. `kernel_size=3` uses a `3 x 3` sliding window.

```python
conv = nn.Conv2d(
    in_channels=3,
    out_channels=6,
    kernel_size=3,
)

output = conv(x)
print(output.shape)
```

#### 4. Padding And Stride

`padding` adds pixels around the image border. `stride` controls how far the kernel moves each step.

```python
# Padding keeps more border information. Stride reduces spatial size faster.
conv2 = nn.Conv2d(
    in_channels=3,
    out_channels=6,
    kernel_size=3,
    padding=1,
    stride=2,
)

output2 = conv2(x)
print(output2.shape)
```

#### Key Idea

A convolution layer changes the image from raw pixels into feature maps. More `out_channels` means more learned feature maps.
