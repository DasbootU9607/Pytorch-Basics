### TensorBoard

TensorBoard is a visualization tool for inspecting training logs, such as scalar curves, images, model graphs, and histograms. In PyTorch, `SummaryWriter` writes event files to a log directory, and the `tensorboard` command reads those event files in the browser.


```python
from torch.utils.tensorboard import SummaryWriter
```

```python
# Optional: inspect the available SummaryWriter methods and arguments.
# This is useful while learning, but not required in normal training code.
help(SummaryWriter)
```

#### 1. Example

```python
writer = SummaryWriter("logs")
```

```python
# writer.add_image()
# y=x
for i in range(100):
    writer.add_scalar("y=x",i,i)
writer.close()
```

**How to open Logs (In WSL Terminal):**

tensorboard --logdir=logs

**To change port number:**

tensorboard --logdir=logs --port=3000
