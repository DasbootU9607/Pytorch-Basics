### DataLoader Basics

This notebook shows how to load CIFAR-10 and wrap it with a PyTorch `DataLoader` for mini-batch iteration.

#### 1. Imports

`torchvision` provides datasets and transforms. `DataLoader` creates batches from a dataset.

```python
# torchvision includes common datasets and image transforms.
import torchvision

# DataLoader helps us read data in mini-batches.
from torch.utils.data import DataLoader

# SummaryWriter writes images and metrics for TensorBoard.
from torch.utils.tensorboard import SummaryWriter
```

#### 2. Create The Dataset

`train=False` loads the test split. `ToTensor()` converts each image from PIL format to a tensor.

```python
test_data = torchvision.datasets.CIFAR10(
    "./dataset",
    train=False,
    transform=torchvision.transforms.ToTensor(),
)
```

#### 3. Build The DataLoader

The loader returns batches of 4 samples. `shuffle=True` changes the sample order each epoch.

```python
test_loader = DataLoader(
    dataset=test_data,
    batch_size=4,
    shuffle=True,
    num_workers=0,
    drop_last=False,
)
```

#### Batch Size Note

A **sample** is one image and its label. A **batch** is a small group of samples returned together by the `DataLoader`.

Here, `batch_size=4`, so each batch contains 4 images. CIFAR-10 test data has 10,000 images, so one full pass gives `10000 / 4 = 2500` batches.

#### 4. Check One Sample

Each CIFAR-10 image becomes a tensor with shape `[channels, height, width]`.

```python
# Get the first image and its class label.
img, target = test_data[0]
print(img.shape)
print(target)
```

#### 5. Log Batches To TensorBoard

This loop reads batches from the `DataLoader` and saves image grids so they can be viewed in TensorBoard.

```python
writer = SummaryWriter("dataloader")

for epoch in range(2):
    step = 0
    for data in test_loader:
        imgs, targets = data
        print(imgs.shape)
        print(targets)

        # Log one mini-batch of images at this step.
        writer.add_images(f"Epoch: {epoch}", imgs, step)
        step += 1

writer.close()
```

#### 6. Open TensorBoard

Run this command in the terminal from the notebook folder:

`tensorboard --logdir="dataloader"`
