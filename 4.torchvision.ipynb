### Torchvision

`torchvision` provides common tools for computer vision tasks.

In this notebook, we use it for three things:

- loading a built-in dataset: `torchvision.datasets.CIFAR10`;
- preprocessing images with `torchvision.transforms`;
- writing sample images to TensorBoard.


#### 1. Import Packages and Define Transform

CIFAR-10 images are PIL images by default. We use `ToTensor()` so each image becomes a PyTorch tensor.


````python
import torchvision
from torch.utils.tensorboard import SummaryWriter

# Compose is useful when we want to apply several transforms in order.
# Here we only need one transform: convert PIL image to tensor.
dataset_transform = torchvision.transforms.Compose([
    torchvision.transforms.ToTensor(),
])
````

#### 2. Load CIFAR-10 Dataset

`CIFAR10` is a small image classification dataset with 10 classes.

Important arguments:

- `root="./dataset"`: save/read the dataset from the local `dataset` folder;
- `train=True`: load the training set;
- `train=False`: load the test set;
- `transform=dataset_transform`: apply the transform when reading each image;
- `download=True`: download the dataset if it is not already available locally.


````python
train_set = torchvision.datasets.CIFAR10(
    root="./dataset",
    train=True,
    transform=dataset_transform,
    download=True,
)

test_set = torchvision.datasets.CIFAR10(
    root="./dataset",
    train=False,
    transform=dataset_transform,
    download=True,
)
````

#### 3. Inspect One Dataset Item

Each item in the dataset returns two values:

```text
(image_tensor, target_label_index)
```

The label index can be converted to the class name with `test_set.classes[target]`.


````python
print(test_set[0])
print(test_set.classes)
````

#### 4. Read Image and Label Separately

This makes the dataset output easier to understand.


````python
img, target = test_set[10]

print(img)                    # Image tensor.
print(target)                 # Class index.
print(test_set.classes[target])  # Class name.
print(img.shape)              # CIFAR-10 tensor shape: C x H x W.
````

#### 5. Write Dataset Images to TensorBoard

Because `ToTensor()` returns images in `C x H x W` format, `writer.add_image()` can read them directly.

Use `test_set[i]` inside the loop so TensorBoard shows different images, not the same image repeated.


````python
writer = SummaryWriter("p4")

for i in range(10):
    img, target = test_set[i]
    writer.add_image("test_set", img, global_step=i)

writer.close()
````

#### 6. Terminal Command

Run this from the same folder as this notebook:

```bash
tensorboard --logdir=p4
```

Then open the URL shown in the terminal, usually:

```text
http://localhost:6006/
```

If port `6006` is already in use:

```bash
tensorboard --logdir=p4 --port=6007
```


#### Checklist

- Use `torchvision.datasets` to load common vision datasets.
- Use `transform=` to preprocess each image when the dataset reads it.
- Use `ToTensor()` before logging images with `SummaryWriter.add_image()`.
- A dataset item usually returns `(image, label)`.
- Use `dataset.classes[label]` to convert a label index into a class name.
